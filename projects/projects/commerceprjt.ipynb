```python
import pandas as pd

customers = pd.read_sql_query(
    "SELECT * FROM customers",
    con=engine
)

print(customers.head())
```

```python
customers = pd.read_sql_query("SELECT * FROM customers", engine)
products = pd.read_sql_query("SELECT * FROM products", engine)
orders = pd.read_sql_query("SELECT * FROM orders", engine)

print("CUSTOMERS")
display(customers.head())

print("PRODUCTS")
display(products.head())

print("ORDERS")
display(orders.head())
```

```python
print("CUSTOMERS")
print(customers.info())

print("PRODUCTS")
print(products.info())

print("ORDERS")
print(orders.info())
```

```python
print("Customers missing values:")
print(customers.isnull().sum())

print("Products missing values:")
print(products.isnull().sum())

print("Orders missing values:")
print(orders.isnull().sum())
```

```python
print("Customer duplicates:", customers.duplicated().sum())

print("Product duplicates:", products.duplicated().sum())

print("Order duplicates:", orders.duplicated().sum())
```

```python
customers_clean = customers.drop_duplicates().copy()

products_clean = products.drop_duplicates().copy()

orders_clean = orders.drop_duplicates().copy()
```

```python
print(customers_clean.isnull().sum())
print(products_clean.isnull().sum())
print(orders_clean.isnull().sum())
```

```python
orders_clean['order_date'] = pd.to_datetime(
    orders_clean['order_date'],
    errors='coerce'
)
```

```python
print(customers_clean.head())
print(products_clean.head())
print(orders_clean.head())

print(orders_clean.dtypes)
```

```python
customers_clean.to_sql(
    name="customers_clean",
    con=engine,
    if_exists="replace",
    index=False
)

print("Customers loaded successfully!")
```

```python
products_clean.to_sql(
    name="products_clean",
    con=engine,
    if_exists="replace",
    index=False
)

print("Products loaded successfully!")
```

```python
orders_clean.to_sql(
    name="orders_clean",
    con=engine,
    if_exists="replace",
    index=False
)

print("Orders loaded successfully!")
```

```python
%%sql
SELECT
    o.order_id,
    c.customer_name,
    p.product_name,
    p.category,
    o.quantity,
    p.unit_price,
    o.quantity * p.unit_price AS total_amount,
    o.order_date
FROM orders_clean AS o
JOIN customers_clean AS c
    ON o.customer_id = c.customer_id
JOIN products_clean AS p
    ON o.product_id = p.product_id;
```

```python
%%sql
SELECT
    p.category,
    SUM(o.quantity * p.unit_price) AS total_sales
FROM orders_clean AS o
JOIN products_clean AS p
    ON o.product_id = p.product_id
GROUP BY p.category
ORDER BY total_sales DESC;
```

```python
%%sql
SELECT
    c.customer_id,
    c.customer_name,
    SUM(o.quantity * p.unit_price) AS total_spending
FROM orders_clean AS o
JOIN customers_clean AS c
    ON o.customer_id = c.customer_id
JOIN products_clean AS p
    ON o.product_id = p.product_id
GROUP BY c.customer_id, c.customer_name
ORDER BY total_spending DESC;
```

```python
%%sql
SELECT
    c.customer_name,
    COUNT(o.order_id) AS total_orders
FROM customers_clean AS c
LEFT JOIN orders_clean AS o
    ON c.customer_id = o.customer_id
GROUP BY c.customer_id, c.customer_name
ORDER BY total_orders DESC;
```

```python
%%sql
SELECT
    DATE(o.order_date) AS order_day,
    SUM(o.quantity * p.unit_price) AS daily_sales
FROM orders_clean AS o
JOIN products_clean AS p
    ON o.product_id = p.product_id
GROUP BY DATE(o.order_date)
ORDER BY order_day;
```

```python
%%sql
SELECT
    p.product_name,
    SUM(o.quantity) AS total_quantity_sold,
    SUM(o.quantity * p.unit_price) AS total_revenue
FROM orders_clean AS o
JOIN products_clean AS p
    ON o.product_id = p.product_id
GROUP BY p.product_id, p.product_name
ORDER BY total_quantity_sold DESC;
```

```python
%%sql
SELECT
    COUNT(*) AS total_order_lines,
    SUM(o.quantity * p.unit_price) AS total_revenue,
    SUM(o.quantity * p.unit_price) / COUNT(DISTINCT o.order_id)
        AS average_order_value
FROM orders_clean AS o
JOIN products_clean AS p
    ON o.product_id = p.product_id;
```

```python
%%sql
CREATE TABLE sales_report AS
SELECT
    o.order_id,
    c.customer_id,
    c.customer_name,
    c.city,
    p.product_id,
    p.product_name,
    p.category,
    o.quantity,
    p.unit_price,
    o.quantity * p.unit_price AS total_amount,
    o.order_date
FROM orders_clean AS o
JOIN customers_clean AS c
    ON o.customer_id = c.customer_id
JOIN products_clean AS p
    ON o.product_id = p.product_id;
```

```python
%%sql
SELECT * FROM sales_report;
```
